```python
import sys
a = input('Введите целое число больше 1: ')
try:
  a = int(a)
except ValueError:
  print ('Введите целое число!') #обработка неправильного ввода

try:
    if int(a) < 2:
     sys.exit()
except SystemExit:
    print ('Введите число больше 1!') #проверка на нужный диапазон
```

```python
all_num = dict() #задаем пустой словарь

for i in range(2,a+1):
  all_num[i] = True #заполняем словарь всеми числами от 2 до введенного как ключами, значения - булевые переменные тру

#print (all_num)
```

```python
for i in range(2,a+1,1): #цикл по алгоритму решета Эратосфена
  if i*i > a:
    break
  if all_num [i] == True:
    for j in range(i*i,a+1,i):
      all_num[j] = False

#print(all_num)
```

```python
for i in range(2,a+1): #вывод прослых чисел, у которых остались зачения тру
  if all_num[i] == True:
    print(i)
```
